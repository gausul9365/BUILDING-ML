```python
from transformers import AutoTokenizer
from transformers import AutoModelForCausalLM
import torch

# loading tokenizer
model_name = "openai-community/gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)


# loading pretrained model Because GPT-2 is a decoder-only causal language mode
model = AutoModelForCausalLM.from_pretrained(model_name)

prompt = "good people"

input = tokenizer(prompt, return_tensors="pt")

# generating new token so
with torch.no_grad():
  outputs = model.generate(
      **input,
      max_new_tokens = 100,
      do_sample=True,
      temperature=0.8,
      top_p=0.9,
      repetition_penalty=1.2
  )


# converting generated token id back to human readable form so
generated_text = tokenizer.decode(
    outputs[0],
    # Passing [0] extracts that tensor row.
    skip_special_tokens=True
)

print(generated_text)
```

What does **model.generate()** do?
This is the most important operation.

It performs autoregressive generation.

The model:
1. Processes the input tokens.
2. Produces vocabulary logits for the next token.
3. Selects a token according to the generation strategy.
4. Appends that token to the sequence.
5. Repeats until a stopping condition is reached.

The model produces logits, softmax converts them into probabilities, and a decoding strategy selects the next token.
