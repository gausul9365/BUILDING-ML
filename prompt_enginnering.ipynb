```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name = "openai-community/gpt2"

model = AutoModelForCausalLM.from_pretrained(model_name)


tokenizer = AutoTokenizer.from_pretrained(model_name)

prompt = """
Classify the sentiment as Positive, Negative, or Neutral.

Review: The product quality is excellent.
Sentiment:
"""

inputs = tokenizer(prompt, return_tensors="pt")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=15,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
prompt = """
Summarize the following text in one sentence.

Text: Deep learning is a subset of machine learning
that uses neural networks with multiple layers to
learn complex patterns from data.

Summary:
"""

inputs = tokenizer(prompt, return_tensors="pt")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=35,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
prompt = """
Answer the question using the provided context.

Context: Python is a programming language widely
used in machine learning and data science.

Question: Which programming language is widely
used in machine learning?

Answer:
"""

inputs = tokenizer(prompt, return_tensors="pt")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=25,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

The problem is that GPT-2 was pretrained primarily for next-token prediction, not specifically optimized to follow instructions or perform summarization.

Prompt =  design	The model may not interpret the instruction as intended.

Training objective = 	GPT-2 learned to continue text, not reliably execute arbitrary instructions.

Training = data	It wasn't specifically instruction-tuned to behave like an assistant.

Architecture =	Causal attention still provides access to the preceding document.

```python
prompt = """
Text: Deep learning is a subset of machine learning
that uses neural networks with multiple layers to
learn complex patterns from data.

A short summary of the text is:
"""

inputs = tokenizer(prompt, return_tensors="pt")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=25,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
